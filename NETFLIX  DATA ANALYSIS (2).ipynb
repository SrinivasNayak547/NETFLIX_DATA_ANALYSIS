```python
df.drop_duplicates(subset='show_id', inplace=True)
```

```python
df['director'] = df['director'].fillna('Unknown')

df['cast'] = df['cast'].fillna('Unknown')

df['country'] = df['country'].fillna('Unknown')

df['rating'] = df['rating'].fillna('Not Rated')
```

```python
df.isnull().sum()
```

```python
df['date_added'] = pd.to_datetime(
    df['date_added'],
    errors='coerce'
)
```

```python
df['type'].value_counts()
```

```python
#Display the content distribution
plt.figure(figsize=(7,5))

sns.countplot(data=df, x='type')

plt.title('Netflix Movies vs TV Shows')
plt.xlabel('Content Type')
plt.ylabel('Total Count')

plt.show()
```

```python
#Which are the top 10 countries producing Netflix content?
countries = df['country'].str.split(', ').explode()

top_countries = countries[
    countries != 'Unknown'
].value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_countries.values,
    y=top_countries.index
)

plt.title('Top 10 Countries on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Country')

plt.show()
```

```python
#Which are the top 10 genres?
genres = df['listed_in'].str.split(', ').explode()

top_genres = genres.value_counts().head(10)

plt.figure(figsize=(10,6))

sns.barplot(
    x=top_genres.values,
    y=top_genres.index
)

plt.title('Top 10 Netflix Genres')
plt.xlabel('Number of Titles')
plt.ylabel('Genre')

plt.show()
```

```python
#Which year had the highest content releases?
yearly_content = df['release_year'].value_counts().sort_index()

yearly_content.plot(
    figsize=(12,5),
    kind='line',
    marker='o'
)

plt.title('Netflix Content Release Trends')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.show()
```

```python
#Which directors have the most Netflix titles?
top_directors = (
    df[df['director'] != 'Unknown']['director']
    .str.split(', ')
    .explode()
    .value_counts()
    .head(10)
)

top_directors.plot(
    kind='barh',
    figsize=(10,6)
)

plt.title('Top 10 Netflix Directors')
plt.xlabel('Number of Titles')
plt.ylabel('Director')

plt.gca().invert_yaxis()
plt.show()
```

```python
#Which content ratings are most common?
plt.figure(figsize=(10,6))

sns.countplot(
    data=df,
    y='rating',
    order=df['rating'].value_counts().head(10).index
)

plt.title('Top Netflix Content Ratings')
plt.xlabel('Number of Titles')
plt.ylabel('Rating')

plt.show()
```

```python
#What is the average movie duration?
movie_duration = df[df['type'] == 'Movie'].copy()

movie_duration['duration_minutes'] = pd.to_numeric(
    movie_duration['duration'].str.extract(r'(\d+)')[0],
    errors='coerce'
)

print(
    'Average movie duration:',
    round(movie_duration['duration_minutes'].mean(), 2),
    'minutes'
)
```

```python
#Which country produces the most movies?
movies = df[df['type'] == 'Movie']

movie_countries = movies['country'].str.split(', ').explode()

print(
    movie_countries[movie_countries != 'Unknown']
    .value_counts()
    .head(10)
)
```

```python
#Which country produces the most TV shows?
tv_shows = df[df['type'] == 'TV Show']

tv_countries = tv_shows['country'].str.split(', ').explode()

print(
    tv_countries[tv_countries != 'Unknown']
    .value_counts()
    .head(10)
)
```

```python
#How many Indian movies and TV shows are available?
india_content = df[
    df['country'].str.contains('India', na=False)
]

print(india_content['type'].value_counts())
```

```python
#Which month has the highest content additions?
df['month_added'] = df['date_added'].dt.month_name()

monthly_content = df['month_added'].value_counts()

monthly_content.plot(
    kind='bar',
    figsize=(10,5)
)

plt.title('Netflix Content Added by Month')
plt.xlabel('Month')
plt.ylabel('Number of Titles')

plt.xticks(rotation=45)
plt.show()
```

```python
#How has Netflix content grown over time?
yearly_added = (
    df['date_added']
    .dropna()
    .dt.year
    .value_counts()
    .sort_index()
)

yearly_added.plot(
    kind='line',
    marker='o',
    figsize=(10,5)
)

plt.title('Netflix Content Addition Trends')
plt.xlabel('Year Added')
plt.ylabel('Number of Titles')

plt.show()
```

```python
#Which actors appear most frequently?
actors = df['cast'].str.split(', ').explode()

top_actors = (
    actors[actors != 'Unknown']
    .value_counts()
    .head(10)
)

top_actors.plot(
    kind='barh',
    figsize=(10,6)
)

plt.title('Top 10 Actors on Netflix')
plt.xlabel('Number of Titles')
plt.ylabel('Actor')

plt.gca().invert_yaxis()
plt.show()
```

```python
plt.figure(figsize=(10,5))

plt.hist(
    movie_duration['duration_minutes'].dropna(),
    bins=30,
    edgecolor='black'
)

plt.title('Netflix Movie Duration Distribution')
plt.xlabel('Duration in Minutes')
plt.ylabel('Number of Movies')

plt.show()
```

```python
#Movies vs TV Shows by release year
plt.figure(figsize=(12,6))

sns.countplot(
    data=df,
    x='release_year',
    hue='type'
)

plt.xlim(0, None)
plt.xticks(rotation=90)

plt.title('Movies vs TV Shows by Release Year')
plt.xlabel('Release Year')
plt.ylabel('Number of Titles')

plt.show()
```

```python
#Content distribution by country
country_type = df[
    df['country'] != 'Unknown'
].copy()

country_type['country'] = country_type['country'].str.split(', ')

country_type = country_type.explode('country')

country_pivot = pd.crosstab(
    country_type['country'],
    country_type['type']
)

top_10 = country_pivot.sum(axis=1).nlargest(10)

top_10_data = country_pivot.loc[top_10.index]

top_10_data.plot(
    kind='bar',
    stacked=True,
    figsize=(12,6)
)

plt.title('Movies and TV Shows by Top 10 Countries')
plt.xlabel('Country')
plt.ylabel('Number of Titles')

plt.xticks(rotation=45)
plt.show()
```
